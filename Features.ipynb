```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.image import load_img, img_to_array

# Define Dice loss function
def dice_loss(y_true, y_pred, smooth=1e-6):
    y_true_f = tf.reshape(y_true, [-1])  # Flatten the tensor
    y_pred_f = tf.reshape(y_pred, [-1])  # Flatten the tensor
    intersection = tf.reduce_sum(y_true_f * y_pred_f)
    union = tf.reduce_sum(y_true_f) + tf.reduce_sum(y_pred_f)
    dice = (2. * intersection + smooth) / (union + smooth)
    return 1 - dice

# Load your saved model
model = load_model("Unet_new_model_run.h5", compile=False)  # Avoid compiling during loading
model.summary()
model.load_weights("Unet_new_model_run.h5")
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Load and preprocess the input image
img_path = "Lung_1400.png"  # Replace with your image path
img = load_img(img_path, target_size=(256, 256), color_mode="grayscale")  # Convert to grayscale
img_array = img_to_array(img) / 255.0  # Normalize the image
img_array = np.expand_dims(img_array, axis=0)  # Add batch dimension

# Target label for the image (replace with actual label if necessary)
target_label = np.zeros((1, 256, 256, 1))  # Example target for binary segmentation

# Gradient logging function for weights
def log_gradients(model, input_image, target_data):
    input_tensor = tf.convert_to_tensor(input_image)  # Convert the input image to a Tensor
    with tf.GradientTape() as tape:
        tape.watch(model.trainable_weights)  # Watch all trainable weights for gradient calculation
        predictions = model(input_tensor, training=True)  # Forward pass
        loss_value = dice_loss(target_data, predictions)  # Use Dice loss for gradient computation

    # Calculate gradients for each trainable weight
    gradients = tape.gradient(loss_value, model.trainable_weights)

    # Create a dictionary of gradients for easier interpretation
    gradient_magnitudes = {}
    for weight, grad in zip(model.trainable_weights, gradients):
        if grad is not None:
            gradient_magnitudes[weight.name] = tf.norm(grad).numpy()  # Calculate L2 norm of the gradient

    return gradient_magnitudes

# Function to plot histograms of gradients
def plot_gradients(gradient_magnitudes, model_name):
    for weight_name, grad_magnitude in gradient_magnitudes.items():
        plt.figure(figsize=(10, 5))
        plt.bar([weight_name], [grad_magnitude])
        plt.title(f"Gradient Magnitudes for {model_name} - {weight_name}")
        plt.ylabel("Gradient Magnitude")
        plt.show()

# Log gradients for your model with the single image
gradients = log_gradients(model, img_array, target_label)

# Plot the gradient magnitudes for visualization
plot_gradients(gradients, "My Model")
```

```python
import tensorflow as tf
from tensorflow.keras.layers import (
    Input, Conv2D, BatchNormalization, Activation, Add, UpSampling2D, Concatenate, Dropout, LeakyReLU, Multiply
)
from tensorflow.keras.models import Model

def ARUNet(input_shape=(256, 256, 1)):
    inputs = Input(shape=input_shape, name="input_layer")

    # Initial block
    conv1 = Conv2D(16, (3, 3), padding="same", name="conv2d_59")(inputs)
    bn1 = BatchNormalization(name="batch_normalization_41")(conv1)
    act1 = Activation("relu", name="activation_19")(bn1)
    skip1 = Conv2D(16, (1, 1), padding="same", name="conv2d_61")(inputs)
    conv2 = Conv2D(16, (3, 3), padding="same", name="conv2d_60")(act1)
    bn2 = BatchNormalization(name="batch_normalization_42")(skip1)
    add1 = Add(name="add_9")([conv2, bn2])
    bn3 = BatchNormalization(name="batch_normalization_43")(add1)
    act2 = Activation("relu", name="activation_20")(bn3)

    # Downsampling block 1
    down1_conv1 = Conv2D(32, (3, 3), strides=2, padding="same", name="conv2d_62")(act2)
    down1_bn1 = BatchNormalization(name="batch_normalization_44")(down1_conv1)
    down1_act1 = Activation("relu", name="activation_21")(down1_bn1)
    skip2 = Conv2D(32, (1, 1), strides=2, padding="same", name="conv2d_64")(add1)
    down1_conv2 = Conv2D(32, (3, 3), padding="same", name="conv2d_63")(down1_act1)
    down1_bn2 = BatchNormalization(name="batch_normalization_45")(skip2)
    down1_add = Add(name="add_10")([down1_conv2, down1_bn2])
    down1_act2 = Activation("relu", name="activation_22")(down1_add)

    # Downsampling block 2
    down2_conv1 = Conv2D(64, (3, 3), strides=2, padding="same", name="conv2d_65")(down1_act2)
    down2_bn1 = BatchNormalization(name="batch_normalization_47")(down2_conv1)
    down2_act1 = Activation("relu", name="activation_24")(down2_bn1)
    skip3 = Conv2D(64, (1, 1), strides=2, padding="same", name="conv2d_67")(down1_add)
    down2_conv2 = Conv2D(64, (3, 3), padding="same", name="conv2d_66")(down2_act1)
    down2_bn2 = BatchNormalization(name="batch_normalization_48")(skip3)
    down2_add = Add(name="add_11")([down2_conv2, down2_bn2])
    down2_act2 = Activation("relu", name="activation_25")(down2_add)

    # Downsampling block 3
    down3_conv1 = Conv2D(128, (3, 3), strides=2, padding="same", name="conv2d_68")(down2_act2)
    down3_bn1 = BatchNormalization(name="batch_normalization_50")(down3_conv1)
    down3_act1 = Activation("relu", name="activation_27")(down3_bn1)
    skip4 = Conv2D(128, (1, 1), strides=2, padding="same", name="conv2d_70")(down2_add)
    down3_conv2 = Conv2D(128, (3, 3), padding="same", name="conv2d_69")(down3_act1)
    down3_bn2 = BatchNormalization(name="batch_normalization_51")(skip4)
    down3_add = Add(name="add_12")([down3_conv2, down3_bn2])
    down3_act2 = Activation("relu", name="activation_28")(down3_add)

    # Downsampling block 4
    down4_conv1 = Conv2D(256, (3, 3), strides=2, padding="same", name="conv2d_71")(down3_act2)
    down4_bn1 = BatchNormalization(name="batch_normalization_53")(down4_conv1)
    down4_act1 = Activation("relu", name="activation_30")(down4_bn1)
    skip5 = Conv2D(256, (1, 1), strides=2, padding="same", name="conv2d_73")(down3_add)
    down4_conv2 = Conv2D(256, (3, 3), padding="same", name="conv2d_72")(down4_act1)
    down4_bn2 = BatchNormalization(name="batch_normalization_54")(skip5)
    down4_add = Add(name="add_13")([down4_conv2, down4_bn2])
    down4_act2 = Activation("relu", name="activation_31")(down4_add)

    # Bottleneck block
    bottleneck_conv1 = Conv2D(256, (3, 3), padding="same", name="conv2d_74")(down4_act2)
    bottleneck_bn1 = BatchNormalization(name="batch_normalization_56")(bottleneck_conv1)
    bottleneck_act1 = Activation("relu", name="activation_33")(bottleneck_bn1)
    bottleneck_conv2 = Conv2D(256, (3, 3), padding="same", name="conv2d_75")(bottleneck_act1)

    # Build the upsampling path
    # Add corresponding upsampling and concatenation blocks for the decoder
    # Each block should align with the skip connections
    # Use LeakyReLU, UpSampling2D, and Concatenate as needed

    # Final block
    outputs = Conv2D(1, (1, 1), activation="sigmoid", name="conv2d_96")(bottleneck_conv2)

    return Model(inputs=inputs, outputs=outputs, name="ARUNet")

# Instantiate the model
model = ARUNet()
model.summary()
```
